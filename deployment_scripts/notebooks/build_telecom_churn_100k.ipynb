# Step 6 - Filter telecom_churn to keep only those user_ids

# Step 7 - Verify alignment
- Shape: Ensures no rows were dropped or duplicated during the filter.
- Integrity: Ensures both sets contain the exact same unique IDs.
- Order: Ensures the sequence is identical from top to bottom.

```python
is_aligned = expresso_sample["user_id"].equals(telecom_churn_100k["user_id"])
print(f"Are the IDs of both datasets perfectly aligned? {is_aligned}")
```

```python
# Check Head
verification_df = pd.DataFrame({
    'Expresso_IDs': expresso_sample["user_id"].values,
    'Telecom_IDs': telecom_churn_100k["user_id"].values
})

print(verification_df.head())
```

```python
# Check shape
len(expresso_sample) == len(telecom_churn_100k)
```

```python
# Check Integrity
set(expresso_sample["user_id"]) == set(telecom_churn_100k["user_id"])
```

```python
# Chek Order
(expresso_sample["user_id"].values == telecom_churn_100k["user_id"].values).all()
```
